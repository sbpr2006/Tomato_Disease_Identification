# Tomato Leaf Disease — MobileNetV2 Transfer Learning

Rebuilt from the original from-scratch CNN notebook. Two changes drive the accuracy jump:

1. **Fixed the data-halving bug** — the original loader used `validation_split=0.5, subset="training"`, which silently discarded half the dataset before the manual train/val/test split ever ran. That's removed here.
2. **Swapped the from-scratch CNN for a frozen, pretrained MobileNetV2 backbone**, trained in two stages: first a new classification head on top of a frozen base, then a short fine-tuning pass that unfreezes the top layers of the base at a much lower learning rate.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
import numpy as np
import matplotlib.pyplot as plt

print(tf.config.list_physical_devices('GPU'))
```

## Load the dataset

**No `validation_split`/`subset` here** — that's what was eating half your data. We load everything once and do the train/val/test split ourselves below, on the *full* set of images.

Image size is `224x224` to match the resolution MobileNetV2's ImageNet weights were trained at (the original notebook used 256x256, which works structurally but 224 gets you the best transfer performance).

```python
DATA_DIR = r"C:\Users\Raze\projects\p3\Dataset"
IMG_SIZE = (256, 256)
BATCH_SIZE = 32

ds = tf.keras.preprocessing.image_dataset_from_directory(
    DATA_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=42,
)

classes = ds.class_names
n_classes = len(classes)
classes
```

## Check class balance

PlantVillage's tomato subset is known to be imbalanced (Yellow Leaf Curl Virus has thousands of images, Mosaic Virus has only a few hundred). We compute per-class counts directly from the folder structure so we can pass `class_weight` to `model.fit()` later — this stops the model from just defaulting to the majority classes on the images it's unsure about.

```python
import pathlib

data_dir = pathlib.Path(DATA_DIR)
counts = {c: len(list((data_dir / c).glob('*'))) for c in classes}
for c, n in counts.items():
    print(f'{c:50s} {n}')

total = sum(counts.values())
class_weight = {
    i: total / (n_classes * counts[c]) for i, c in enumerate(classes)
}
class_weight
```

## Split train/val/test

Same 80/10/10 split logic as the original notebook, just now running against the full, un-halved dataset.

```python
def get_data_partitions(ds, train_split=0.8, val_split=0.1, shuffle=True, shuffle_size=10000, seed=12):
    ds_size = len(ds)
    train_size = int(train_split * ds_size)
    val_size = int(val_split * ds_size)
    if shuffle:
        ds = ds.shuffle(shuffle_size, seed=seed)
    train_ds = ds.take(train_size)
    remainder = ds.skip(train_size)
    val_ds = remainder.take(val_size)
    test_ds = remainder.skip(val_size)
    return train_ds, val_ds, test_ds

train_ds, val_ds, test_ds = get_data_partitions(ds)
print(f'train batches: {len(train_ds)}, val batches: {len(val_ds)}, test batches: {len(test_ds)}')
```

```python
train_ds = train_ds.cache("train_cache").shuffle(200).prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.cache("val_cache").prefetch(buffer_size=tf.data.AUTOTUNE)
test_ds = test_ds.cache("test_cache").prefetch(buffer_size=tf.data.AUTOTUNE)
```

## Augmentation

Kept from the original notebook. Note there's **no manual `Rescaling(1./255)` layer** this time — `preprocess_input` below handles MobileNetV2's expected `[-1, 1]` input range, and stacking a `1/255` rescale in front of it would double-scale the images and quietly hurt accuracy.

```python
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.2),
    layers.RandomZoom(0.2),
    layers.RandomContrast(0.2),
])
```

## Build the model — frozen MobileNetV2 base + custom head

`include_top=False` drops MobileNetV2's original 1000-class ImageNet head. `base_model.trainable = False` freezes every pretrained weight so stage 1 only trains the new classification head — this is what lets a ~13k-image dataset work well with a network that has millions of parameters, since we're not trying to relearn low-level filters (edges, textures) from scratch.

```python
base_model = MobileNetV2(input_shape=IMG_SIZE + (3,), include_top=False, weights="imagenet")
base_model.trainable = False

inputs = tf.keras.Input(shape=IMG_SIZE + (3,))
x = data_augmentation(inputs)
x = preprocess_input(x)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.2)(x)
outputs = layers.Dense(n_classes, activation="softmax")(x)

model = tf.keras.Model(inputs, outputs)
model.summary()
```

## Compile

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False),
    metrics=["accuracy"],
)
```

```python
callbacks = [
    EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True),
    ReduceLROnPlateau(monitor="val_loss", factor=0.2, patience=2),
    ModelCheckpoint("best_head.weights.h5", monitor="val_accuracy", save_best_only=True, save_weights_only=True),
]
```

## Stage 1 — train the head, base frozen

Expect this to climb fast and plateau somewhere in the low-to-mid 90s — the base already knows how to see edges, textures, and shapes, so the head is really just learning to map MobileNetV2's features onto your 10 disease classes.

```python
history_head = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=12,
    class_weight=class_weight,
    callbacks=callbacks,
)
```

## Stage 2 — fine-tune the top of the base

Now unfreeze the *last* block of MobileNetV2 (leave the early layers — edges, colors, textures — frozen, since those are generic and don't need to change) and continue training at a much lower learning rate. A high LR here would wreck the pretrained weights in a couple of steps.

```python
base_model.trainable = True

fine_tune_at = len(base_model.layers) - 40  # unfreeze roughly the last 40 layers
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False),
    metrics=["accuracy"],
)

fine_tune_callbacks = [
    EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True),
    ReduceLROnPlateau(monitor="val_loss", factor=0.2, patience=2),
    ModelCheckpoint("best_finetuned.weights.h5", monitor="val_accuracy", save_best_only=True, save_weights_only=True),
]

history_finetune = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    class_weight=class_weight,
    callbacks=fine_tune_callbacks,
)
```

## Evaluate on the held-out test set

```python
test_loss, test_acc = model.evaluate(test_ds)
print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_acc:.4f}")
```

## Training curves

Both stages plotted together, with a marker where fine-tuning kicks in.

```python
acc = history_head.history["accuracy"] + history_finetune.history["accuracy"]
val_acc = history_head.history["val_accuracy"] + history_finetune.history["val_accuracy"]
loss = history_head.history["loss"] + history_finetune.history["loss"]
val_loss = history_head.history["val_loss"] + history_finetune.history["val_loss"]
switch_epoch = len(history_head.history["accuracy"])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(acc, label="train")
axes[0].plot(val_acc, label="val")
axes[0].axvline(switch_epoch, color="gray", linestyle="--", label="fine-tune starts")
axes[0].set_title("Accuracy")
axes[0].legend()

axes[1].plot(loss, label="train")
axes[1].plot(val_loss, label="val")
axes[1].axvline(switch_epoch, color="gray", linestyle="--", label="fine-tune starts")
axes[1].set_title("Loss")
axes[1].legend()

plt.show()
```

## Sample predictions

```python
def predict(model, img):
    img_array = tf.expand_dims(img, 0)
    pred = model.predict(img_array, verbose=0)
    pred_class = classes[np.argmax(pred[0])]
    confidence = round(100 * np.max(pred[0]), 2)
    return pred_class, confidence

plt.figure(figsize=(15, 15))
for image_batch, label_batch in test_ds.take(1):
    for i in range(9):
        ax = plt.subplot(3, 3, i + 1)
        plt.imshow(image_batch[i].numpy().astype("uint8"))

        predicted_class, confidence = predict(model, image_batch[i])
        actual_label = classes[label_batch[i].numpy()]

        color = "green" if predicted_class == actual_label else "red"
        plt.title(f"Actual: {actual_label}\nPredicted: {predicted_class} ({confidence:.2f}%)", color=color)
        plt.axis("off")
```
